##Step 2 — Prepare the data loaders

```python

import os
import torch
from torch.utils.data import DataLoader
from torchvision import datasets, transforms

# Dataset paths
BASE_DIR = "/content/drive/MyDrive/PCOS_Project"
SPLIT_DIR = os.path.join(BASE_DIR, "SPLIT_DATASET")
TRAIN_DIR = os.path.join(SPLIT_DIR, "train")
TEST_DIR = os.path.join(SPLIT_DIR, "test")

# Device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# ImageNet normalization (same as ResNet50)
imagenet_mean = [0.485, 0.456, 0.406]
imagenet_std = [0.229, 0.224, 0.225]

# Training transformations
train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(5),
    transforms.ColorJitter(brightness=0.15, contrast=0.15),
    transforms.ToTensor(),
    transforms.Normalize(imagenet_mean, imagenet_std)
])

# Testing transformations: no random augmentation
test_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(imagenet_mean, imagenet_std)
])

# Load datasets
train_dataset = datasets.ImageFolder(
    TRAIN_DIR, transform=train_transform
)
test_dataset = datasets.ImageFolder(
    TEST_DIR, transform=test_transform
)

# Data loaders
train_loader = DataLoader(
    train_dataset, batch_size=32,
    shuffle=True, num_workers=2
)
test_loader = DataLoader(
    test_dataset, batch_size=32,
    shuffle=False, num_workers=2
)

# Verify
print("Class mapping:", train_dataset.class_to_idx)
print("Training images:", len(train_dataset))
print("Testing images:", len(test_dataset))
print("Training batches:", len(train_loader))
print("Testing batches:", len(test_loader))

images, labels = next(iter(train_loader))
print("Image batch shape:", images.shape)
print("Label batch shape:", labels.shape)
```

##Step 3 — Load pretrained EfficientNetB0

```python

import torch
import torch.nn as nn
from torchvision import models

# Confirm device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# Load ImageNet-pretrained EfficientNetB0
weights = models.EfficientNet_B0_Weights.DEFAULT
model = models.efficientnet_b0(weights=weights)

# Replace the final classification layer
num_features = model.classifier[1].in_features
model.classifier[1] = nn.Linear(num_features, 2)

# Move model to GPU
model = model.to(device)

print("Model: EfficientNetB0")
print("Pretrained weights: ImageNet")
print("Input features to final layer:", num_features)
print("Output classes:", model.classifier[1].out_features)
print("Model loaded successfully.")
```

##Step 4 — Configure the loss function and optimizer

```python

import torch.nn as nn
import torch.optim as optim

# Loss function
criterion = nn.CrossEntropyLoss()

# Optimizer
optimizer = optim.Adam(
    model.parameters(),
    lr=0.0001
)

print("Loss function:", criterion)
print("Optimizer:", optimizer.__class__.__name__)
print("Learning rate:", optimizer.param_groups[0]["lr"])
print(
    "Trainable parameters:",
    sum(p.numel() for p in model.parameters() if p.requires_grad)
)
print("Training configuration ready.")
```

##Step 5 — Train baseline EfficientNetB0

```python

import time
import torch

NUM_EPOCHS = 5
train_history = []

for epoch in range(NUM_EPOCHS):
    model.train()

    running_loss = 0.0
    correct = 0
    total = 0
    start_time = time.time()

    for batch_idx, (images, labels) in enumerate(train_loader):
        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        running_loss += loss.item() * images.size(0)
        predictions = outputs.argmax(dim=1)
        correct += (predictions == labels).sum().item()
        total += labels.size(0)

        if (batch_idx + 1) % 10 == 0 or batch_idx + 1 == len(train_loader):
            print(
                f"Epoch {epoch + 1}/{NUM_EPOCHS} | "
                f"Batch {batch_idx + 1}/{len(train_loader)} | "
                f"Loss so far: {running_loss / total:.4f}",
                flush=True
            )

    epoch_loss = running_loss / total
    epoch_accuracy = 100 * correct / total
    elapsed = (time.time() - start_time) / 60

    train_history.append({
        "epoch": epoch + 1,
        "loss": epoch_loss,
        "accuracy": epoch_accuracy
    })

    print(
        f"COMPLETED Epoch [{epoch + 1}/{NUM_EPOCHS}] | "
        f"Loss: {epoch_loss:.4f} | "
        f"Accuracy: {epoch_accuracy:.2f}% | "
        f"Time: {elapsed:.2f} min\n",
        flush=True
    )

print("EfficientNetB0 baseline training completed.")
```

##Step 6 — Save the trained baseline model

```python

import os
import torch

SAVE_DIR = "/content/drive/MyDrive/PCOS_Project/model_checkpoints"
os.makedirs(SAVE_DIR, exist_ok=True)

checkpoint_path = os.path.join(
    SAVE_DIR, "efficientnetb0_baseline.pth"
)

torch.save({
    "model_state_dict": model.state_dict(),
    "class_to_idx": train_dataset.class_to_idx,
    "epochs_trained": 5,
    "image_size": 224
}, checkpoint_path)

print("EfficientNetB0 baseline saved successfully!")
print("Saved to:", checkpoint_path)
print(
    "File size (MB):",
    round(os.path.getsize(checkpoint_path) / (1024 * 1024), 2)
)
```

##Step 7: Evaluate EfficientNetB0 baseline

```python
import torch
import numpy as np
from torchvision import models
from torch import nn
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)

# Rebuild the EfficientNetB0 architecture
model_eval = models.efficientnet_b0(weights=None)
num_features = model_eval.classifier[1].in_features
model_eval.classifier[1] = nn.Linear(num_features, 2)

# Load the saved baseline checkpoint
checkpoint_path = (
    "/content/drive/MyDrive/PCOS_Project/"
    "model_checkpoints/efficientnetb0_baseline.pth"
)

checkpoint = torch.load(
    checkpoint_path,
    map_location=device,
    weights_only=True
)

model_eval.load_state_dict(checkpoint["model_state_dict"])
model_eval = model_eval.to(device)
model_eval.eval()

# Evaluate on the existing test dataset
y_true = []
y_pred = []
y_prob = []

with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)
        outputs = model_eval(images)

        probabilities = torch.softmax(outputs, dim=1)
        predictions = torch.argmax(outputs, dim=1)

        y_true.extend(labels.cpu().numpy())
        y_pred.extend(predictions.cpu().numpy())
        y_prob.extend(probabilities[:, 1].cpu().numpy())

# Calculate metrics
accuracy = accuracy_score(y_true, y_pred)
cm = confusion_matrix(y_true, y_pred, labels=[0, 1])

print("EfficientNetB0 Baseline — Test Results")
print("=" * 45)
print(f"Test images: {len(y_true)}")
print(f"Accuracy: {accuracy * 100:.2f}%")

print("\nClassification Report:")
print(classification_report(
    y_true,
    y_pred,
    labels=[0, 1],
    target_names=["infected", "noninfected"],
    digits=4,
    zero_division=0
))

print("Confusion Matrix:")
print(cm)

print(
    "\nROC-AUC (noninfected as positive class):",
    round(roc_auc_score(y_true, y_prob), 4)
)
```

##Step 9: Load the baseline and prepare class weights

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import models

# Load a separate copy of the baseline model
weighted_model = models.efficientnet_b0(weights=None)
num_features = weighted_model.classifier[1].in_features
weighted_model.classifier[1] = nn.Linear(num_features, 2)

checkpoint_path = (
    "/content/drive/MyDrive/PCOS_Project/"
    "model_checkpoints/efficientnetb0_baseline.pth"
)

checkpoint = torch.load(
    checkpoint_path,
    map_location=device,
    weights_only=True
)

weighted_model.load_state_dict(checkpoint["model_state_dict"])
weighted_model = weighted_model.to(device)

# Calculate class weights using the training dataset
class_counts = torch.bincount(
    torch.tensor(train_dataset.targets, dtype=torch.long)
).float()

class_weights = (
    class_counts.sum() / (len(class_counts) * class_counts)
).to(device)

# Weighted loss and optimizer
criterion_weighted = nn.CrossEntropyLoss(
    weight=class_weights
)

optimizer_weighted = optim.Adam(
    weighted_model.parameters(),
    lr=0.0001
)

print("Model loaded successfully.")
print("Training class counts:", class_counts.tolist())
print("Class weights:", class_weights.cpu().tolist())
print("Loss function:", criterion_weighted)
print("Optimizer: Adam")
print("Learning rate: 0.0001")
```

##Step 10: Train the class-weighted model

```python
import time
import torch

num_epochs = 5

for epoch in range(num_epochs):
    weighted_model.train()

    running_loss = 0.0
    correct = 0
    total = 0
    start_time = time.time()

    for batch_idx, (images, labels) in enumerate(train_loader):
        images = images.to(device)
        labels = labels.to(device)

        optimizer_weighted.zero_grad()

        outputs = weighted_model(images)
        loss = criterion_weighted(outputs, labels)

        loss.backward()
        optimizer_weighted.step()

        running_loss += loss.item() * images.size(0)
        predictions = outputs.argmax(dim=1)
        correct += (predictions == labels).sum().item()
        total += labels.size(0)

        if (batch_idx + 1) % 10 == 0 or (batch_idx + 1) == len(train_loader):
            print(
                f"Epoch [{epoch + 1}/{num_epochs}] "
                f"Batch [{batch_idx + 1}/{len(train_loader)}]"
            )

    epoch_loss = running_loss / total
    epoch_accuracy = 100 * correct / total
    elapsed = (time.time() - start_time) / 60

    print(
        f"Epoch [{epoch + 1}/{num_epochs}] "
        f"Loss: {epoch_loss:.4f} | "
        f"Accuracy: {epoch_accuracy:.2f}% | "
        f"Time: {elapsed:.2f} min"
    )

print("\nEfficientNetB0 class-weighted training completed!")
```

#GRAD-CAM Implementation

##Step 1 — Prepare Grad-CAM for one ultrasound image

```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
import random
import os

# Grad-CAM target layer: final convolutional feature block
target_layer = model.features[-1]

# Store feature maps and gradients
activations = {}
gradients = {}

def save_activation(module, inputs, output):
    activations["value"] = output

def save_gradient(module, grad_input, grad_output):
    gradients["value"] = grad_output[0]

# Register hooks
activation_hook = target_layer.register_forward_hook(save_activation)
gradient_hook = target_layer.register_full_backward_hook(save_gradient)

# Select one test image
image_path, true_label = random.choice(test_dataset.samples)

# Load and preprocess image using the existing test transform
original_image = Image.open(image_path).convert("RGB")
input_tensor = test_dataset.transform(original_image).unsqueeze(0).to(device)

# Forward pass
model.zero_grad()
output = model(input_tensor)
probabilities = torch.softmax(output, dim=1)
predicted_class = output.argmax(dim=1).item()

# Backpropagate the predicted class score
score = output[0, predicted_class]
score.backward()

# Extract activations and gradients
feature_maps = activations["value"].detach()
gradient_maps = gradients["value"].detach()

# Global-average-pool gradients to obtain channel weights
weights = gradient_maps.mean(dim=(2, 3), keepdim=True)

# Weighted sum of feature maps
cam = (weights * feature_maps).sum(dim=1, keepdim=True)
cam = F.relu(cam)

# Resize heatmap to input image size and normalize
cam = F.interpolate(
    cam,
    size=(224, 224),
    mode="bilinear",
    align_corners=False
)
cam = cam[0, 0].cpu().numpy()
cam = (cam - cam.min()) / (cam.max() - cam.min() + 1e-8)

# Display the original image and Grad-CAM heatmap
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(original_image.resize((224, 224)))
plt.title(f"Original image | True class: {test_dataset.classes[true_label]}")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(original_image.resize((224, 224)))
plt.imshow(cam, cmap="jet", alpha=0.45)
plt.title(
    f"Grad-CAM | Predicted: {test_dataset.classes[predicted_class]}\n"
    f"Confidence: {probabilities[0, predicted_class].item() * 100:.2f}%"
)
plt.axis("off")

plt.tight_layout()
plt.show()

# Remove hooks when finished
activation_hook.remove()
gradient_hook.remove()
```

### Grad-CAM Visualization and Interpretation

Gradient-weighted Class Activation Mapping (Grad-CAM) was implemented using the final convolutional layer of the EfficientNetB0 model to visualize image regions that influence its classification prediction. The generated heatmap was overlaid on the original ultrasound image to provide a visual explanation of the model's output.

In the displayed example, the model predicted the PCOS-positive class (`infected`) with a confidence score of 99.99%, and the ground-truth dataset label was also PCOS-positive. The heatmap highlights regions that contributed to the prediction, with warmer colors indicating stronger relative activation.

Grad-CAM provides a visual interpretation of the model's behavior, but it does not independently verify the presence of PCOS or establish clinical validity. Further evaluation on appropriately validated clinical data is required before any diagnostic use.The colored overlay highlights image regions that contributed to the model's prediction. Red/yellow indicates stronger activation relative to the displayed heatmap, while blue indicates weaker activation.
